```python
# ============================================================
# EXPLAINABLE AI (XAI) FOR HEALTHCARE PREDICTIVE MODELS
# ============================================================
#
# Dataset:
# UCI Heart Disease Dataset - Cleveland
#
# Models:
#   1. Logistic Regression
#   2. Random Forest
#   3. Support Vector Machine
#   4. XGBoost
#
# Explainability:
#   1. SHAP
#   2. LIME
#
# Analysis:
#   - Model performance (AUC, F1, Precision, Recall)
#   - Global feature importance
#   - Local patient-level explanations
#   - SHAP vs LIME comparison
#   - Explanation stability (Spearman Rank Correlation)
#   - Feature-ranking agreement
#
# ============================================================

# ============================================================
# 0. INSTALL PACKAGES
# ============================================================
# Uncomment the line below if running in a fresh environment
!pip -q install shap lime xgboost joblib

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================
import os
import random
import warnings

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report,
    roc_curve
)

import shap
from lime.lime_tabular import LimeTabularExplainer
import joblib

warnings.filterwarnings("ignore")

# ============================================================
# 2. REPRODUCIBILITY
# ============================================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
print("Random seed:", SEED)

# ============================================================
# 3. CREATE PROJECT DIRECTORIES
# ============================================================
directories = ["data", "plots", "models", "results", "explanations"]
for directory in directories:
    os.makedirs(directory, exist_ok=True)
print("Project directories created.")

# ============================================================
# 4. ENVIRONMENT-AGNOSTIC DATA LOADING
# ============================================================
# This block allows the code to run seamlessly whether you are
# presenting it via Google Colab or executing it locally.
# It references the file named verbatim processed.cleveland.data.

cleveland_file = "/content/processed.cleveland.data"

if not os.path.exists(cleveland_file):
    try:
        from google.colab import files
        print(f"Running in Google Colab. Please upload '{cleveland_file}'.")
        uploaded = files.upload()
    except ImportError:
        raise FileNotFoundError(f"File '{cleveland_file}' not found. Please place it in the working directory.")
else:
    print(f"Found '{cleveland_file}' locally.")

# ============================================================
# 5. COLUMN DEFINITIONS
# ============================================================
column_names = [
    "age", "sex", "cp", "trestbps", "chol", "fbs", "restecg",
    "thalach", "exang", "oldpeak", "slope", "ca", "thal", "target"
]

# ============================================================
# 6. LOAD CLEVELAND DATASET
# ============================================================
# The dataset uses '?' to denote missing values.
df = pd.read_csv(
    cleveland_file,
    header=None,
    names=column_names,
    na_values="?",
    skipinitialspace=True
)

print("\nDataset loaded successfully. Shape:", df.shape)

# ============================================================
# 7. DUPLICATE CHECK
# ============================================================
duplicate_count = df.duplicated().sum()
if duplicate_count > 0:
    df = df.drop_duplicates()
    print(f"\nRemoved {duplicate_count} duplicate rows. New shape: {df.shape}")

# ============================================================
# 8. CONVERT DATA TO NUMERIC
# ============================================================
for column in df.columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# ============================================================
# 9. RIGOROUS MISSING VALUE IMPUTATION
# ============================================================
# Methodological refinement: Use median imputation for continuous
# variables and mode imputation for categorical/ordinal variables
# (e.g., ca and thal have missing values).
categorical_cols = ["sex", "cp", "fbs", "restecg", "exang", "slope", "ca", "thal"]
continuous_cols = ["age", "trestbps", "chol", "thalach", "oldpeak"]

for column in df.columns:
    if column != "target":
        if column in categorical_cols:
            mode_value = df[column].mode()[0]
            df[column] = df[column].fillna(mode_value)
        elif column in continuous_cols:
            median_value = df[column].median()
            df[column] = df[column].fillna(median_value)

# Target rows with missing targets are dropped to ensure label integrity
df = df.dropna(subset=["target"])
print("\nMissing values remaining after imputation:\n", df.isnull().sum())

# ============================================================
# 10. CONVERT TARGET TO BINARY
# ============================================================
# Clinical Justification: The dataset target ranges from 0 to 4[cite: 2].
# 0 indicates no disease, while 1, 2, 3, and 4 indicate varying severity of disease[cite: 2].
# We map >0 to 1 to formulate a binary "presence vs. absence" task,
# providing a cleaner decision boundary for XAI evaluation.
df["target"] = (df["target"] > 0).astype(int)

# ============================================================
# 11. SAVE CLEAN DATASET
# ============================================================
df.to_csv("data/heart_disease_cleveland_clean.csv", index=False)

# ============================================================
# 12. EDA - TARGET DISTRIBUTION & CORRELATION
# ============================================================
df.describe().to_csv("results/descriptive_statistics.csv")

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="target")
plt.title("Heart Disease Class Distribution")
plt.xlabel("Heart Disease (0 = Absent, 1 = Present)")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.savefig("plots/class_distribution.png", dpi=300)
plt.close()

plt.figure(figsize=(12, 9))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Matrix of Clinical Variables")
plt.tight_layout()
plt.savefig("plots/correlation_matrix.png", dpi=300)
plt.close()

# ============================================================
# 13. TRAIN / TEST SPLIT
# ============================================================
X = df.drop(columns=["target"])
y = df["target"]

# Stratified split to maintain class proportions
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=SEED, stratify=y
)

# ============================================================
# 14. MODEL DEFINITIONS (PREVENTING DATA LEAKAGE)
# ============================================================
# Standard scaling is embedded inside pipelines for linear/kernel
# models to compute statistics strictly on the training folds.
models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=2000, random_state=SEED))
    ]),
    "Random Forest": RandomForestClassifier(
        n_estimators=300, max_depth=8, min_samples_split=5, random_state=SEED, n_jobs=-1
    ),
    "SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC(probability=True, kernel="rbf", random_state=SEED))
    ]),
    "XGBoost": XGBClassifier(
        n_estimators=300, max_depth=4, learning_rate=0.03, subsample=0.8,
        colsample_bytree=0.8, eval_metric="logloss", random_state=SEED, n_jobs=-1
    )
}

# ============================================================
# 15. TRAIN & SAVE MODELS
# ============================================================
trained_models = {}
for name, model in models.items():
    model.fit(X_train, y_train)
    trained_models[name] = model
    filename = name.lower().replace(" ", "_") + ".joblib"
    joblib.dump(model, f"models/{filename}")

# ============================================================
# 16. MODEL EVALUATION
# ============================================================
model_results = []
predictions, probabilities = {}, {}

for name, model in trained_models.items():
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    predictions[name] = y_pred
    probabilities[name] = y_prob

    model_results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1": f1_score(y_test, y_pred, zero_division=0),
        "ROC_AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(model_results).sort_values("ROC_AUC", ascending=False)
results_df.to_csv("results/model_performance.csv", index=False)
print("\n=== MODEL PERFORMANCE ===")
print(results_df)

# ============================================================
# 17. PERFORMANCE BAR CHART
# ============================================================
performance_plot = results_df.set_index("Model")[["Accuracy", "Precision", "Recall", "F1", "ROC_AUC"]]
performance_plot.plot(kind="bar", figsize=(12, 6))
plt.title("Predictive Performance Comparison")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.ylim(0, 1.05)
plt.legend(loc='lower right')
plt.tight_layout()
plt.savefig("plots/model_performance_comparison.png", dpi=300)
plt.close()

# ============================================================
# 18. ROC CURVES
# ============================================================
plt.figure(figsize=(10, 7))
for name in trained_models.keys():
    fpr, tpr, _ = roc_curve(y_test, probabilities[name])
    auc = roc_auc_score(y_test, probabilities[name])
    plt.plot(fpr, tpr, label=f"{name} (AUC={auc:.3f})")

plt.plot([0, 1], [0, 1], linestyle="--", color='grey')
plt.title("ROC Curves")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()
plt.savefig("plots/roc_curves.png", dpi=300)
plt.close()

# ============================================================
# 19. SHAP - RANDOM FOREST (GLOBAL)
# ============================================================
print("\nCalculating SHAP values for Random Forest...")
rf_model = trained_models["Random Forest"]
rf_explainer = shap.TreeExplainer(rf_model)
rf_shap_values = rf_explainer.shap_values(X_test)

def get_positive_class_shap(shap_vals):
    if isinstance(shap_vals, list): return shap_vals[1]
    shap_vals = np.asarray(shap_vals)
    if shap_vals.ndim == 3: return shap_vals[:, :, 1]
    return shap_vals

rf_shap_positive = get_positive_class_shap(rf_shap_values)

shap.summary_plot(rf_shap_positive, X_test, show=False)
plt.title("SHAP Summary - Random Forest")
plt.tight_layout()
plt.savefig("plots/shap_random_forest_summary.png", dpi=300, bbox_inches="tight")
plt.close()

rf_shap_importance = pd.DataFrame({
    "Feature": X_test.columns,
    "Mean_Absolute_SHAP": np.mean(np.abs(rf_shap_positive), axis=0)
}).sort_values("Mean_Absolute_SHAP", ascending=False)
rf_shap_importance.to_csv("results/shap_random_forest_feature_importance.csv", index=False)

# ============================================================
# 20. SHAP - RANDOM FOREST (LOCAL EXPLANATION)
# ============================================================
sample_index = 0
sample = X_test.iloc[sample_index]
expected_value = rf_explainer.expected_value
base_value = expected_value[1] if isinstance(expected_value, np.ndarray) else expected_value

patient_explanation = shap.Explanation(
    values=rf_shap_positive[sample_index],
    base_values=base_value,
    data=sample.values,
    feature_names=X.columns.tolist()
)

shap.plots.waterfall(patient_explanation, show=False)
plt.tight_layout()
plt.savefig("plots/shap_local_patient.png", dpi=300, bbox_inches="tight")
plt.close()

# ============================================================
# 21. LIME EXPLAINER & LOCAL EXPLANATION
# ============================================================
print("\nCreating LIME explainer...")
lime_explainer = LimeTabularExplainer(
    training_data=X_train.values,
    feature_names=X_train.columns.tolist(),
    class_names=["No Disease", "Disease"],
    mode="classification",
    discretize_continuous=False,
    random_state=SEED
)

lime_explanation = lime_explainer.explain_instance(
    X_test.iloc[sample_index].values,
    rf_model.predict_proba,
    num_features=len(X.columns)
)

fig = lime_explanation.as_pyplot_figure()
plt.title("LIME Local Explanation - Random Forest")
plt.tight_layout()
plt.savefig("plots/lime_random_forest_patient.png", dpi=300, bbox_inches="tight")
plt.close()

# ============================================================
# 22. AGGREGATED LIME IMPORTANCE
# ============================================================
N_LIME_SAMPLES = min(50, len(X_test))
lime_records = []

for i in range(N_LIME_SAMPLES):
    explanation = lime_explainer.explain_instance(
        X_test.iloc[i].values, rf_model.predict_proba, num_features=len(X.columns)
    )
    for feature, weight in explanation.as_list():
        lime_records.append({"Sample": i, "Feature": feature, "Weight": weight})

lime_multiple_df = pd.DataFrame(lime_records)
lime_global = (
    lime_multiple_df.assign(Absolute_Weight=lambda x: np.abs(x["Weight"]))
    .groupby("Feature")["Absolute_Weight"].mean().reset_index()
    .sort_values("Absolute_Weight", ascending=False)
)

# ============================================================
# 23. SHAP VS LIME AGREEMENT & SPEARMAN RANK CORRELATION
# ============================================================
shap_compare = rf_shap_importance.rename(columns={"Mean_Absolute_SHAP": "SHAP_Importance"})
lime_compare = lime_global.rename(columns={"Absolute_Weight": "LIME_Importance"})

importance_comparison = pd.merge(shap_compare, lime_compare, on="Feature", how="outer").fillna(0)
importance_comparison["SHAP_Rank"] = importance_comparison["SHAP_Importance"].rank(ascending=False)
importance_comparison["LIME_Rank"] = importance_comparison["LIME_Importance"].rank(ascending=False)

rank_correlation = importance_comparison[["SHAP_Rank", "LIME_Rank"]].corr(method="spearman").iloc[0, 1]

TOP_N = 10
top_shap_features = set(rf_shap_importance.head(TOP_N)["Feature"])
top_lime_features = set(lime_global.head(TOP_N)["Feature"])
common_features = top_shap_features.intersection(top_lime_features)
feature_agreement = len(common_features) / TOP_N

print(f"\nSHAP vs LIME Spearman rank correlation: {rank_correlation:.4f}")
print(f"Top-10 feature agreement: {feature_agreement:.4f}")

# ============================================================
# 24. SHAP EXPLANATION STABILITY ANALYSIS
# ============================================================
print("\nCalculating SHAP explanation stability...")
N_STABILITY_REPEATS = 5
STABILITY_SAMPLE_SIZE = min(100, len(X_test))
rng = np.random.default_rng(SEED)

shap_rankings = []
for repeat in range(N_STABILITY_REPEATS):
    indices = rng.choice(len(X_test), size=STABILITY_SAMPLE_SIZE, replace=False)
    subset_values = rf_shap_positive[indices]
    feature_importance = np.mean(np.abs(subset_values), axis=0)
    ranking = pd.Series(feature_importance, index=X_test.columns).rank(ascending=False)
    ranking.name = f"Repeat_{repeat + 1}"
    shap_rankings.append(ranking)

shap_stability_corr = pd.concat(shap_rankings, axis=1).corr(method="spearman")
upper_triangle = np.triu_indices_from(shap_stability_corr, k=1)
shap_stability_score = shap_stability_corr.values[upper_triangle].mean()

print(f"Overall SHAP stability score: {shap_stability_score:.4f}")

# ============================================================
# 25. FINAL OUTPUT & PIPELINE SUMMARY
# ============================================================
best_model_name = results_df.iloc[0]["Model"]
best_auc = results_df.iloc[0]["ROC_AUC"]

with open("results/project_summary.txt", "w") as file:
    file.write("EXPLAINABLE AI FOR HEALTHCARE PREDICTIVE MODELS\n")
    file.write("=" * 60 + "\n\n")
    file.write("Dataset: UCI Heart Disease Dataset - Cleveland\n")
    file.write("Task: Binary heart disease classification (0 = Absent, >0 = Present)\n\n")
    file.write(f"Best predictive model: {best_model_name}\n")
    file.write(f"Best ROC-AUC: {best_auc:.4f}\n")
    file.write(f"SHAP mathematical stability score: {shap_stability_score:.4f}\n")
    file.write(f"SHAP-LIME rank correlation: {rank_correlation:.4f}\n")
    file.write(f"Top-10 feature agreement: {feature_agreement:.4f}\n\n")
    file.write("Common top features:\n")
    for feature in sorted(common_features):
        file.write(f"- {feature}\n")

print("\n=======================================================")
print("PROJECT PIPELINE COMPLETED SUCCESSFULLY")
print("=======================================================")
print(f"Best Model: {best_model_name} (AUC: {best_auc:.4f})")
print("All artifacts, plots, and models saved to respective directories.")
```
